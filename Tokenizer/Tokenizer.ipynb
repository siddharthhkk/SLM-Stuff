````python
def get_state(token):
    counts = {}
    for pair in zip(token, token[1:]):  #iterate over consecutive elements
        counts[pair] = counts.get(pair, 0) + 1
    return counts
stats = get_state(tokens)
print(sorted(((v,k) for k, v in stats.items()), reverse=True)) # value how many times that key pair occured
chr(101), chr(32)
````

````python
# now we have identified the most common pair we want to invent a new token with the id of 256 as ascii already have 0-255!
# iterating over the tokens and whenever we see (101, 32) we swap it out with 256!
top_pair: list = max(stats, key=stats.get)

def merge(token, pair, idx):
    # replace all occurences of pair with new token idx
    new_tokens = []
    i = 0
    while i < len(token):
        # if not at the very last position and pair matches, replace it
        if i < len(token) - 1 and token[i] == pair[0] and token[i+1] == pair[1]:
            new_tokens.append(idx)
            i += 2
        else:
            new_tokens.append(token[i])
            i += 1
    return new_tokens

new_tokens = merge(tokens, top_pair, 256)
print(len(new_tokens), len(tokens)) # new_token should be 20 less than tokens
````

It has its own training dataset of text (which could be different from that of the LLM), on which you train the vocabulary using the Byte Pair Encoding (BPE) algorithm. It then translates back and forth between raw text and sequences of tokens. The LLM later only ever sees the tokens and never directly deals with any text.

````python
# decoding - given a sequence of integers in range [0, vocab_size] whats the text?
vocab = {idx : bytes([idx]) for idx in range(256)}
for (p0, p1), idx in merges.items():
    vocab[idx] = vocab[p0] + vocab[p1]  #integers to the bytes

def decode(ids: list):
    tokens = b"".join(vocab[idx] for idx in ids)
    text = tokens.decode('utf-8', errors='replace')
    return text
print(decode([275]))


# encoding - given a string we want to get the tokens, in our vocab including the merges we did!
def encode(text):
    tokens = list(text.encode('utf-8')) # tokens in utf8 vocab of size 256 
    while len(tokens) > 2 : # to handle single character edge case
        stats = get_state(tokens)
        pair = min(stats, key= lambda p: merges.get(p, float('inf')))   # min as we want the longest match first, as we want to replace the most frequent pair with a new token
        if pair not in merges:
            break  # nothing left to merge
        idx = merges[pair]
        tokens = merge(tokens, pair, idx)
        return tokens
print(encode('hello world'))

#checks 
print(decode(encode("hello world")))
test = decode(encode(text2))
print(test == text2)
````

sentencepiece

Commonly used because (unlike tiktoken) it can efficiently both train and inference BPE tokenizers. It is used in both Llama and Mistral series.

The big difference: sentencepiece runs BPE on the Unicode code points directly! It then has an option character_coverage for what to do with very very rare codepoints that appear very few times, and it either maps them onto an UNK token, or if byte_fallback is turned on, it encodes them with utf-8 and then encodes the raw bytes instead.

TLDR:
    tiktoken encodes to utf-8 and then BPEs bytes
    sentencepiece BPEs the code points and optionally falls back to utf-8 bytes for rare code points (rarity is determined by character_coverage hyperparameter), which then get translated to byte tokens.


````python
import sentencepiece as spm
import os
# as it requires a txt file lets first write it 
with open('toy400.txt', 'w', encoding='utf-8') as f:
    f.write("SentencePiece is an unsupervised text tokenizer and detokenizer mainly for Neural Network-based text generation systems where the vocabulary size is predetermined prior to the neural model training. SentencePiece implements subword units (e.g., byte-pair-encoding (BPE) [Sennrich et al.]) and unigram language model [Kudo.]) with the extension of direct training from raw sentences. SentencePiece allows us to make a purely end-to-end system that does not depend on language-specific pre/postprocessing.")


spm.SentencePieceTrainer.train(
    input = 'toy400.txt',
    input_format = 'text',
    model_prefix = 'token400',  #output filename
    model_type = 'bpe',
    vocab_size = 400,
    # we want as raw data as possible so no normalization!
    normalization_rule_name = 'identity',
    remove_extra_whitespaces=False,
    input_sentence_size = 200000000, # max number of training sentences
    max_sentence_length=4192, # max number of bytes per sentence
    seed_sentencepiece_size=1000000,
    shuffle_input_sentence=True,
    # rare word treatment
    character_coverage=0.99995,
    byte_fallback=False,              #byte_fallback
    # merge rules
    split_digits=True,
    split_by_unicode_script=True,
    split_by_whitespace=True,
    split_by_number=True,
    max_sentencepiece_length=16,
    add_dummy_prefix=True,            #add_dummy_prefix
    allow_whitespace_only_pieces=True,
    # special tokens
    unk_id = 0,   # for the unk token
    bos_id = 1,   # beginning of sentence token
    eos_id = 2,   # end of sentence token
    pad_id = -1,   # -1 to turn off
    num_threads=os.cpu_count(),   #use all cpu cores!
)
````

````python
sp = spm.SentencePieceProcessor()
sp.load('token400.model')
vocab = [[sp.id_to_piece(idx), idx] for idx in range(sp.get_piece_size())]
print(f'vocab: {vocab}')   
# you wil see how <unk> is at 0 then <s> for bos is 1 and </s> for eos is 2 and then comes byte_fallback from 3 to 258 and after that the tokens which it has learned!
````

````python
token = sp.encode("hello, 안녕하세요")
print(token)
print([sp.id_to_piece(idx) for idx in token])
# as we can see ut-of-vocabulary characters like "안녕하세요" are split into UTF-8 byte tokens, if i turn byte_fallbacks to false then they are just <unk> tokens.
# and add_dummy_prefix adds a space '_' before the first char like ' h' this is helpful as if some tokens are saved as ' h' and a char char of text is jsut 'h',
# that will be a different token so we do this to have same token for first character aswell
````
